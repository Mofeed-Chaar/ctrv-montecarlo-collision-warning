```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt
import copy
```

```python
image_path = "sample/white_image-1.png"
print(os.path.isfile(image_path))
```

```python
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_size = img.shape
height = img_size[1]
width = img_size[0]
white_image = np.ones((height, width, 3), dtype=np.uint8) * 255
#plt.imshow(img)
```

```python
def put_area(img,color,rectangle,text):
    cv2.rectangle(img, rectangle[0], rectangle[1], color, 3)
    print((rectangle[0][0]+rectangle[1][0])/2)

    cv2.putText(img, text, (int((rectangle[0][0]+rectangle[1][0])/2)-60,int((rectangle[0][1]+rectangle[1][1])/2)), cv2.FONT_HERSHEY_SIMPLEX, 1.5, color, 3)
```

```python
ar1_a = (850, 450)
ar1_b = (1050, 650)
copy_img = copy.deepcopy(img)
cv2.rectangle(copy_img, ar1_a, ar1_b, (255, 0, 0), 3)

ar2_a = (650, 550)
ar2_b = (850, 780)
cv2.rectangle(copy_img, ar2_a, ar2_b, (255, 0, 0), 3)
ar3_a = (450, 550)
ar3_b = (650, 780)
cv2.rectangle(copy_img, ar3_a, ar3_b, (255, 0, 0), 3)

plt.imshow(copy_img)
```

```python
put_area(white_image,(255, 0, 0),(ar1_a, ar1_b),'Area 1')
put_area(white_image,(255, 0, 0),(ar2_a, ar2_b),'Area 2')
put_area(white_image,(255, 0, 0),(ar3_a, ar3_b),'Area 3')
plt.imshow(white_image)
```
